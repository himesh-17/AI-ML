```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,accuracy_score ,precision_score
```

```python
df = pd.read_csv("loan_approval_data.csv")
df.head()
df.info()
```

Handle Missing Value

```python
categorical_data = df.select_dtypes(["object"]).columns
numerical_data = df.select_dtypes(["number"]).columns
```

```python
from sklearn.impute import SimpleImputer

num_imputer = SimpleImputer(strategy="mean")
df[numerical_data] = num_imputer.fit_transform(df[numerical_data])

cat_imputer  = SimpleImputer(strategy="most_frequent")
df[categorical_data] = cat_imputer.fit_transform(df[categorical_data])
```

EDA - exploratory data analysis

```python
classes_count = df["Loan_Approved"].value_counts()

plt.pie(classes_count , labels=["No" , "Yes"] , autopct="%1.1f%%")
plt.title("Loan approved or not")
```

```python
gender_count = df["Gender"].value_counts()
ax = sns.barplot(gender_count)
ax.bar_label(ax.containers[0])
```
